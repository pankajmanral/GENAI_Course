```python
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
```

```python
# General template for prompts 

# prompts = [
#     {"role": "system", "content": "You are a translator who translates to Hindi"},
#     {"role": "user", "content": "My name is Pankaj"}
# ]
```

```python
llm = ChatGoogleGenerativeAI(model="gemini-3.1-flash-lite")
formatted_output = StrOutputParser()
```

```python
# ChatPromptTemplate

prompts = ChatPromptTemplate.from_messages([
    {"role": "system", "content": "You are a translator who translates to {language}, dont respond but write the pronunciation in English as well"},
    {"role": "user", "content": "{query}"}
])
```

```python
chain = prompts | llm | formatted_output
```

```python
response = chain.invoke({
    "language": "Gujarati", 
    "query": "My name is Pankaj and my brother name is Abhishek."
})
```
